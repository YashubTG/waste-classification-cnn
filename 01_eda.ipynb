```python
import os
dataset_path = "dataset-resized"
classes = os.listdir(dataset_path)
print(classes)
print(len(classes))
```

```python
classes = [
    item for item in os.listdir(dataset_path)
    if os.path.isdir(os.path.join(dataset_path, item))
]

print(classes)
print(len(classes))
```

```python
for class_name in classes:
    class_path = os.path.join(dataset_path, class_name)
    images = os.listdir(class_path)
    print(class_name, ":", len(images))
```

```python
from PIL import Image
import matplotlib.pyplot as plt
```

```python
fig, axes = plt.subplots(2, 3, figsize=(12, 8))

for ax, class_name in zip(axes.ravel(), classes):
    class_path = os.path.join(dataset_path, class_name)
    image_name = os.listdir(class_path)[0]
    image_path = os.path.join(class_path, image_name)

    image = Image.open(image_path)

    ax.imshow(image)
    ax.set_title(class_name)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(2,3, figsize=(12,8))

for ax, item in zip(axes.ravel(), classes):
    class_path = os.path.join(dataset_path,item)
    image_name = os.listdir(class_path)[1]
    image_path = os.path.join(class_path, image_name)

    image = Image.open(image_path)
    ax.imshow(image)
    ax.set_title(item)
    ax.axis("off")
plt.tight_layout()
plt.show()


```

```python
for class_name in classes:
    class_path = os.path.join(dataset_path, class_name)
    image1 = os.listdir(class_path)[0]
    image_path1 = os.path.join(class_path, image1)
    image1 = Image.open(image_path1)
    ref_img = image1.size
    images = os.listdir(class_path)
    image_count = len(images)
    for c in range(image_count):
        image_name = images[c]
        image_path = os.path.join(class_path, image_name)
        image = Image.open(image_path)
        if ref_img == image.size:
            state = "all other image sizes in this class is same as reference class"
            continue
        else:
            print(class_name,":", image.size)
    print(class_name , ":" ,state )


```
